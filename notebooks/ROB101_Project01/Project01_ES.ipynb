### Matrices

Sea $A$ una matriz de dimensiones $n\times m$:

- `size(A,1)` devuelve $n$, el número de filas de la matriz.
- `size(A,2)` devuelve $m$, el número de columnas de la matriz.
- `size(A)` devuelve el par $(n,m)$, indicando primero el número de filas y luego el número de columnas.
- `length(A)` devuelve $n \times m$, es decir, el número total de elementos contenidos en la matriz.

### Vectores

Sea $v$ un vector de dimensión $n$:

- `length(v)` devuelve $n$, el número de elementos del vector.
- Aunque es posible aplicar la función `size()` a un vector, generalmente no es la opción recomendada.
- Para vectores, se prefiere utilizar `length()` porque expresa directamente la cantidad de elementos que contiene.

### Importante

En la práctica, rara vez se necesita conocer el número total de elementos de una matriz.

La mayoría de las operaciones requieren conocer:

- el número de filas y columnas;
- únicamente el número de filas; o
- únicamente el número de columnas.

Por este motivo, las funciones `size(A,1)` y `size(A,2)` suelen ser mucho más útiles que `length(A)` cuando se trabaja con matrices.

```julia
# Sumar un vector compatible a CADA una de las columnas de una matriz
#
# A continuación se muestran tres métodos para sumar el vector v3
# a cada columna de la matriz A.

C = copy(A)

# Método 1: fuerza bruta
C[:,1] = C[:,1] + v3
C[:,2] = C[:,2] + v3
C[:,3] = C[:,3] + v3
C[:,4] = C[:,4] + v3

display(A)
display(C)
display(C - A)

# La matriz A tiene tamaño 3×4 y el vector v3 tiene dimensión 3.
#
# Cada columna de A es un vector de dimensión 3:
#
# A = [a₁ a₂ a₃ a₄]
#
# donde:
#
# a₁, a₂, a₃ y a₄ ∈ R³
#
# Al sumar v3 a cada columna obtenemos:
#
# C = [a₁+v3  a₂+v3  a₃+v3  a₄+v3]
#
# Observe que:
#
# C - A
#
# devuelve una matriz cuyas columnas son exactamente v3.
#
# Este ejemplo ilustra una idea fundamental del Álgebra Lineal:
# una misma operación vectorial puede aplicarse simultáneamente
# a muchas columnas de una matriz.
#
# Más adelante utilizaremos esta misma idea para trasladar miles
# o millones de puntos de una nube LiDAR mediante una única
# operación matricial.
```

```julia
C = copy(A)

# Método 2: utilizar un bucle for
for i = 1:size(C,2)
    C[:,i] = C[:,i] + v3   # C[:,i] selecciona todas las filas y la columna i de C
end

display(A)
display(C)
display(C - A)

# En lugar de escribir una instrucción para cada columna,
# utilizamos un bucle for para recorrer automáticamente
# todas las columnas de la matriz.
#
# size(C,2) devuelve el número de columnas de C.
#
# Si C es una matriz 3×4:
#
# i = 1, 2, 3, 4
#
# En cada iteración se suma el vector v3 a la columna i:
#
# C[:,i] = C[:,i] + v3
#
# donde:
#
# C[:,i]
#
# significa:
#
# ":"  -> todas las filas
# "i"  -> columna i
#
# Este enfoque es más escalable que el Método 1.
#
# Si la matriz tuviera 100 columnas, el mismo código
# seguiría funcionando sin modificaciones.
#
# En aplicaciones reales de robótica y procesamiento
# de datos es común trabajar con miles o millones de
# vectores almacenados como columnas de una matriz.
#
# Por ello, automatizar operaciones mediante bucles
# constituye una herramienta fundamental.
#
# Observe nuevamente que:
#
# C - A
#
# contiene el efecto de la transformación aplicada,
# es decir, el vector v3 sumado a cada columna.
```

```julia
# Este es el método más compacto y normalmente el más utilizado en Julia.
#
# La operación:
#
# C .+ v3
#
# indica que Julia debe sumar el vector v3 elemento a elemento
# sobre toda la matriz C utilizando broadcasting.
#
# Conceptualmente es equivalente a:
#
# for i = 1:size(C,2)
#     C[:,i] = C[:,i] + v3
# end
#
# pero Julia realiza la operación de manera más eficiente y con
# una sintaxis mucho más compacta.
#
# El punto (.) es fundamental.
#
# Correcto:
#
# C .+ v3
#
# Incorrecto:
#
# C + v3
#
# Sin el punto, Julia intentará realizar una suma matricial
# convencional entre objetos con dimensiones incompatibles,
# produciendo un error.
#
# Desde el punto de vista geométrico, estamos trasladando todas
# las columnas de la matriz mediante el mismo vector v3.
#
# Esta idea aparece constantemente en:
#
# - procesamiento de imágenes;
# - visión artificial;
# - robótica móvil;
# - aprendizaje automático;
# - procesamiento de nubes de puntos LiDAR.
#
# Más adelante utilizaremos exactamente este principio para
# trasladar simultáneamente miles o millones de puntos mediante
# operaciones matriciales vectorizadas.
```

```julia
# Utilice su método favorito para sumar el vector v4 a cada columna
# de la matriz B.
#
# Guarde el resultado en la matriz C.

C = copy(B)

# Método recomendado: broadcasting
C = C .+ v4

# Las verificaciones asumen que el resultado está almacenado en C
E = C - B; test = true
for i = 1:length(v4)
    for j = 1:size(E,2)
        test = test & isapprox(E[i,j], i, atol = 1e-2)
    end
end

if test == true
    println("Buen trabajo")
else
    println("Inténtelo nuevamente. Necesitará dominar esta operación en el Proyecto 1.")
end

# B es una matriz de tamaño 4×10.
#
# v4 es un vector de dimensión 4:
#
# v4 = [1; 2; 3; 4]
#
# Como cada columna de B tiene exactamente 4 elementos,
# podemos sumar v4 a todas las columnas simultáneamente.
#
# Geométricamente estamos trasladando cada columna de B
# mediante el mismo vector v4.
#
# La operación:
#
# C = C .+ v4
#
# utiliza broadcasting para aplicar la misma suma a todas
# las columnas de la matriz.
#
# Esta idea será utilizada más adelante para transformar
# grandes conjuntos de puntos mediante operaciones matriciales
# eficientes.
```

```julia
# Multiplicar una matriz compatible por cada columna de una segunda matriz
#
# A continuación se muestran distintas formas de multiplicar cada
# columna de la matriz B por la matriz A y almacenar los resultados
# en una matriz C de dimensiones adecuadas.


# Método 1: utilizando un bucle for

nRowsA = size(A,1)
nColsB = size(B,2)

C = zeros(nRowsA, nColsB)   # matriz de ceros del tamaño correcto

for k = 1:size(B,2)
    v = B[:,k]      # extraer la columna k de B
    Av = A * v      # multiplicar A por la columna k
    C[:,k] = Av     # almacenar el resultado en la columna k de C
end


# Método 1-a: una pequeña variación del método anterior

D = zeros(nRowsA, nColsB)   # matriz de ceros del tamaño correcto

for k = 1:size(B,2)
    D[:,k] = A * B[:,k]     # realizar todo en una sola instrucción
end


# Método 1-b: otra variación

E = Array{Float64}(undef, nRowsA, nColsB)  # matriz vacía del tamaño correcto
                                           # creada utilizando undef

for k = 1:size(B,2)
    E[:,k] = A * B[:,k]     # realizar todo en una sola instrucción
end

# ¿Son iguales los resultados?

display(C - D)
display(C - E)

# Recordemos las dimensiones:
#
# A es una matriz 3×4
# B es una matriz 4×10
#
# Por lo tanto:
#
# A * B
#
# es una multiplicación válida porque las dimensiones internas
# coinciden:
#
# (3×4)(4×10) = (3×10)
#
# En este ejemplo estamos construyendo manualmente el producto
# matricial columna por columna.
#
# Si:
#
# B = [b₁ b₂ b₃ ... b₁₀]
#
# entonces:
#
# AB = [Ab₁ Ab₂ Ab₃ ... Ab₁₀]
#
# Cada columna de la matriz resultado se obtiene multiplicando
# A por la columna correspondiente de B.
#
# Los tres métodos calculan exactamente el mismo resultado:
#
# C = AB
# D = AB
# E = AB
#
# Por ello:
#
# C - D = 0
# C - E = 0
#
# donde 0 representa una matriz de ceros.
#
# Este ejemplo es importante porque muestra una interpretación
# fundamental del producto matricial:
#
# "Multiplicar una matriz por otra equivale a multiplicar la
# primera matriz por cada columna de la segunda."
#
# Esta interpretación será utilizada posteriormente para
# transformar simultáneamente grandes conjuntos de puntos
# mediante una única multiplicación matricial.
```

### Parte B: Traslación como multiplicación matricial en coordenadas homogéneas 

Nuestro objetivo aquí es mostrar cómo la operación de traslación puede escribirse como una multiplicación matriz-vector. Los especialistas en robótica desean hacer esto para que las tres operaciones —traslación, rotación y escalamiento— puedan escribirse exactamente con el mismo formato. Este pequeño trabajo adicional facilita mucho la manipulación de archivos LiDAR de gran tamaño.

Planteamos la siguiente pregunta: ¿cómo podemos usar la multiplicación matricial para lograr una traslación? La respuesta es que necesitamos utilizar un truco llamado **coordenadas homogéneas**. Esta es una operación simple sobre nuestros datos, porque solo necesitamos agregar un 1 a cada punto $[x;y]$. Es decir, convertimos $\begin{bmatrix} x \\ y \end{bmatrix}$ a coordenadas homogéneas mediante $$\begin{bmatrix} x \\ y \\ 1 \end{bmatrix}.$$

La operación inversa, tomar un vector en coordenadas homogéneas y eliminar el 1, produce el vector original $[x;y]$. Esto parece casi demasiado simple como para mencionarlo, ¡pero lo hacemos de todos modos!

Afirmamos que podemos realizar una traslación vectorial multiplicando un vector en coordenadas homogéneas por la matriz $$ T = \begin{bmatrix} 1 & 0 & t_x \\ 0 & 1 & t_y \\ 0 & 0 & 1 \end{bmatrix}.$$

Veamos si esto realmente funciona $$ \begin{bmatrix} 1 & 0 & t_x \\ 0 & 1 & t_y \\ 0 & 0 & 1 \end{bmatrix} * \begin{bmatrix} x \\ y \\ 1 \end{bmatrix} =  \begin{bmatrix} 1 \\ 0 \\ 0 \end{bmatrix} *  x +  \begin{bmatrix} 0 \\ 1 \\ 0 \end{bmatrix} * y +  \begin{bmatrix} t_x \\ t_y \\ 1 \end{bmatrix} * 1 = \begin{bmatrix} x + t_x \\ y + t_y \\ 1 \end{bmatrix},$$

donde hemos utilizado nuestro método de multiplicación matricial de columnas por filas. Puede rehacerlo usted mismo usando el método clásico de multiplicación matriz por vector para obtener el mismo resultado. Y cuando eliminamos el 1 para volver a colocar nuestro vector transformado en coordenadas cartesianas ordinarias, obtenemos

$$ \begin{bmatrix} x + t_x \\ y + t_y \end{bmatrix},$$

¡Listo! Está bien si esta operación no le resulta impresionante. Nos gusta construir el suspenso poco a poco.

### Parte E: Integrándolo todo: Transformaciones Afines (3 %)

Todas las transformaciones que hemos realizado hasta ahora —traslación, rotación y escalamiento—, así como sus combinaciones, se denominan **transformaciones afines**.

Cuando trabajamos con grandes conjuntos de datos y necesitamos aplicar una secuencia de traslaciones, rotaciones y escalamientos a un conjunto de puntos, puede imaginarse que llevar el control de todo el proceso de transformación puede volverse complicado con el tiempo.

La ventaja principal de utilizar la noción abstracta de una transformación afín es que podemos construir una transformación compleja multiplicando entre sí las transformaciones individuales consideradas una a una.

Por ejemplo, supongamos que queremos trasladar un conjunto de puntos por `[-0.5; -0.5]`, luego rotar el objeto trasladado 45 grados en sentido antihorario y, finalmente, escalar el objeto ya trasladado y rotado. ¡Son muchas transformaciones! $$T = \begin{bmatrix}1.5&0&0\\0&0.5&0\\0&0&1\\ \end{bmatrix} \cdot \begin{bmatrix}\cos(\pi/4)&-\sin(\pi/4)&0\\\sin(\pi/4)&\cos(\pi/4)&0\\0&0&1\\ \end{bmatrix} \cdot \begin{bmatrix}1&0&-0.5\\0&1&-0.5\\0&0&1\\ \end{bmatrix}$$

¿Qué hemos hecho? Preparamos la transformación homogénea total multiplicando tres transformaciones homogéneas individuales, que luego podemos aplicar todas de una sola vez a nuestro punto `[x;y]`, una vez que lo expresamos en coordenadas homogéneas como `[x;y;1]`.

```julia
# Ejecute esta celda. No debe modificarla.
#
# Esta sección contiene funciones auxiliares que serán utilizadas
# durante toda la Tarea 3.
#
# Léala con atención:
#
#  • muestra cómo organizar código reutilizable;
#  • introduce el concepto de funciones;
#  • ilustra cómo procesar grandes volúmenes de datos;
#  • proporciona un ejemplo práctico del uso de transformaciones afines
#    sobre nubes de puntos LiDAR.
#
# En el Laboratorio 4 aprenderá a construir funciones similares.

using LinearAlgebra
using DelimitedFiles
using Plots
default(fmt = :png)

# ============================================================
# FUNCIÓN data_parser
# ============================================================
#
# Esta función carga y organiza automáticamente todos los datos
# asociados a un intervalo temporal de la trayectoria de Cassie.
#
# Entrada:
#
#     id ∈ {9,10,11,12,13}
#
# donde id identifica el segundo del recorrido que deseamos analizar.
#
# Salida:
#
#     pointcloud_data  -> coordenadas xyz de la nube de puntos
#     Intensity_data   -> intensidad LiDAR de cada punto
#     R                -> matriz de rotación
#     t                -> vector de traslación
#     pose             -> posición y orientación real del robot
#
# El objetivo es ocultar la complejidad de leer múltiples archivos
# CSV y permitir que el estudiante se concentre en el Álgebra Lineal.

function data_parser(id)

    # --------------------------------------------------------
    # Carga de la nube de puntos LiDAR
    # --------------------------------------------------------

    points = readdlm(
        string("data/cassie_data/Frame_",
               string(id),
               "/pointcloud.csv"), ',')

    pointcloud_data = points[1:3,:]    # coordenadas xyz
    Intensity_data = points[4,:]       # intensidad LiDAR

    # --------------------------------------------------------
    # Carga de las transformaciones geométricas
    # --------------------------------------------------------
    #
    # Estas matrices representan la transformación entre
    # distintos sistemas de referencia.
    #
    # Relacione este concepto con las transformaciones afines
    # estudiadas en las Tareas 1 y 2.

    Transformations = readdlm(
        string("data/cassie_data/Frame_",
               string(id),
               "/Transformations.csv"), ',')

    R = Transformations[:,1:3]   # matriz de rotación
    t = Transformations[:,4]     # vector de traslación

    # --------------------------------------------------------
    # Carga de la pose real del robot
    # --------------------------------------------------------
    #
    # Pose = posición + orientación

    pose = readdlm(
        string("data/cassie_data/Frame_",
               string(id),
               "/Position.csv"), ',')

    return (pointcloud_data,
            Intensity_data,
            R,
            t,
            pose)
end


# ============================================================
# EJEMPLO DE USO DEL PARSER
# ============================================================
#
# Se cargan los datos correspondientes al instante t = 10 s.

(pointcloud_data,
 Intensity_data,
 R,
 T,
 position) = data_parser(10)


# ============================================================
# FUNCIÓN affine_transform
# ============================================================
#
# Esta función aplica una transformación homogénea a una nube
# de puntos 3D.
#
# Observe que es exactamente el mismo procedimiento utilizado
# en las Tareas 1 y 2:
#
#   1. Homogeneizar los puntos agregando una fila de unos.
#   2. Aplicar una transformación afín.
#   3. Eliminar la coordenada homogénea.
#
# La diferencia es que ahora trabajamos con miles o cientos
# de miles de puntos simultáneamente.
#
# Esta es una de las ventajas fundamentales del Álgebra Lineal:
#
#     una sola multiplicación matricial
#
# puede transformar enormes conjuntos de datos.

function affine_transform(points, H)

    # Homogeneización
    transformed_p = [copy(points);
                     ones(1,size(points,2))]

    # Transformación afín
    transformed_p = H * transformed_p

    # Deshomogeneización
    return transformed_p[1:3,:]
end


# ============================================================
# FUNCIÓN robot_arrow
# ============================================================
#
# Función auxiliar utilizada únicamente para visualizar
# la orientación del robot en las gráficas.
#
# Extrae la dirección de avance a partir del ángulo yaw.
#
# No es necesario comprender los detalles internos para
# resolver este proyecto.

function robot_arrow(pose)

    position = pose[:,1]

    (roll,pitch,yaw) = pose[:,2]

    # Vector unitario asociado a la orientación del robot
    u = [1*cos(yaw-pi/2)]
    v = [1*sin(yaw-pi/2)]

    return (position, u, v)
end


# ============================================================
# PARÁMETROS DE VISUALIZACIÓN
# ============================================================
#
# Definen el tamaño de la ventana utilizada para graficar
# la nube de puntos y la trayectoria del robot.

window_size = 15

min_x = -window_size
max_x =  window_size

min_y = -window_size
max_y =  window_size

min_z = 0
max_z = 2*window_size

limits = (
    (min_x, max_x),
    (min_y, max_y),
    (min_z, max_z)
)
```
